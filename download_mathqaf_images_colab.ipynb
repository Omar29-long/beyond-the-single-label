# Download mathqaf images

Télécharge les images référencées dans `mathqaf_media_full.csv` (3475 posts).

**Important** : les URLs sont des liens CDN Instagram signés (tokens à durée limitée, ~24-48h). Lance ce notebook rapidement après avoir exporté le CSV.

Étape 1 : uploade `mathqaf_media_full.csv` (celui dans ton dossier Téléchargements) quand la cellule ci-dessous te le demande.

```python
from google.colab import files

uploaded = files.upload()  # sélectionne mathqaf_media_full.csv
csv_path = list(uploaded.keys())[0]
print(f"Fichier chargé : {csv_path}")
```

```python
import csv
import os
import time

import requests

OUTDIR = "images"
DELAY_SECONDS = 0.1
TIMEOUT = 15

os.makedirs(OUTDIR, exist_ok=True)

with open(csv_path, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

print(f"{len(rows)} lignes chargées depuis le CSV")
```

```python
total = len(rows)
ok = 0
skipped = 0
failed = 0
failed_codes = []

for i, row in enumerate(rows, start=1):
    code = (row.get("code") or "").strip()
    url = (row.get("image_url") or "").strip()

    if not code or not url:
        skipped += 1
        continue

    out_path = os.path.join(OUTDIR, f"{code}.jpg")

    if os.path.exists(out_path) and os.path.getsize(out_path) > 0:
        skipped += 1
        continue

    try:
        resp = requests.get(url, timeout=TIMEOUT)
        resp.raise_for_status()
        with open(out_path, "wb") as img_f:
            img_f.write(resp.content)
        ok += 1
    except Exception as e:
        failed += 1
        failed_codes.append(code)

    if i % 200 == 0:
        print(f"Progress: {i}/{total} (ok={ok}, skipped={skipped}, failed={failed})")

    time.sleep(DELAY_SECONDS)

print("\nTerminé.")
print(f"  Téléchargées : {ok}")
print(f"  Déjà présentes/URL manquante : {skipped}")
print(f"  Échouées : {failed}")
if failed_codes:
    print(f"  Codes en échec (premiers 20): {failed_codes[:20]}")
```

## Zipper et télécharger le dossier images

```python
import shutil

shutil.make_archive("mathqaf_images", "zip", OUTDIR)
files.download("mathqaf_images.zip")
```
